## Cleaning & Structring the data

Issues in file
1. inconsistant data (mixed types) - 'rating'
2. missing data (age)
3. duplicate data (same name repeated)

```python
import json
```

```python
# load the file(json)
def load_data(filename):
    with open(filename,'r') as f:
        data = json.load(f)
        
    return data
```

```python
# call function
data = load_data("data.json")
print(data, "\n",type(data))
```

```python
# clean and struct the data
def clean_data(data):
    nums = {"one": 1,"two": 2,"three": 3,"four": 4,"five": 5,"six": 6,"seven": 7,"eight": 8,"nine": 9,"ten":10,"zero":0}
    cleaned_data = []
    unique_set = set()
    # clean the rates
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if (raw_rating in nums):
            raw_rating = nums[raw_rating]
        user['rating'] = raw_rating

        # assigning the empty/missing values with 'None'
        raw_age = user.get("age")
        if (raw_age == None):
            user["age"] = None
    
        #deduplication
        if(user["age"] in unique_set):
            continue
        unique_set.add(user["age"])
        
        #remove extra spaces in name
        if user.get("name") != None:
            user["name"] = user["name"].strip()
        
        
        cleaned_data.append(user)
        
        
        
    return cleaned_data
```

```python
clean_data(data)
```

## Generating meaning insights from data

```python
data = clean_data(data)
```

```python
def generate_insights(data):

    # avg of rating
    total_rate = 0
    poor_rate = 0
    for user in data:
        total_rate += float(user["rating"])
        if float(user["rating"]) < 3:
            poor_rate += 1
   
    avg = total_rate/len(data)
    print("Average rating: ",avg)
    # % members rated poor (<3)
    percent = (poor_rate/len(data))*100
    print("% of members gave less than 3 rating: ",percent,"%")
```

```python
generate_insights(data)
```

## Recommendation System

1. if rate >=4 , recommend same type products.(ex: apple)
2. if rate < 4 , recommend diffrent product(ex:samsung)

```python
def get_recommendation(data):

    recommendation = []

    for user in data:
        curent_recomm = {}
        curent_recomm['name'] = user['name']

        if float(user['rating']) >= 4:
            curent_recomm['brand'] = 'Apple'
        else:
            curent_recomm['brand'] = 'Samsung'

        recommendation.append(curent_recomm)

    return recommendation
```

```python
get_recommendation(data)
```
