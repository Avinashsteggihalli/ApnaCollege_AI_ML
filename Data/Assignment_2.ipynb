## Cleaning & Structring the data

Issues in file
1. inconsistant data (mixed types) - 'rating'
2. missing data (age)
3. duplicate data (same name repeated)

```python
import json
```

```python
# load the file(json)
def load_data(filename):
    with open(filename,'r') as f:
        data = json.load(f)
        
    return data
```

```python
# call function
data = load_data("data.json")
print(data, "\n",type(data))
```

```python
# clean and struct the data
def clean_data(data):
    nums = {"one": 1,"two": 2,"three": 3,"four": 4,"five": 5,"six": 6,"seven": 7,"eight": 8,"nine": 9,"ten":10,"zero":0}
    cleaned_data = []
    unique_set = set()
    # clean the rates
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if (raw_rating in nums):
            raw_rating = nums[raw_rating]
        user['rating'] = raw_rating

        # assigning the empty/missing values with 'None'
        raw_age = user.get("age")
        if (raw_age == None):
            user["age"] = None
    
        #deduplication
        if(user["age"] in unique_set):
            continue
        unique_set.add(user["age"])
        
        #remove extra spaces in name
        if user.get("name") != None:
            user["name"] = user["name"].strip()
        
        
        cleaned_data.append(user)
        
        
        
    return cleaned_data
```

```python
clean_data(data)
```
