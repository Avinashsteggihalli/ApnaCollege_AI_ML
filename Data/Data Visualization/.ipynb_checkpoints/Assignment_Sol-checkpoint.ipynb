## Q1. For the following dataset generated using NumPy:

    np.random.normal(loc=70, scale=10, size=200)

* Plot a **histogram** of scores  
* Draw a vertical line for the mean & try to interpret the results

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
ds = np.random.normal(loc=70, scale=10, size=200)
ds
```

```python
# Plot a histogram of scores
plt.hist(ds,edgecolor='black')
plt.xlabel("Score")
plt.show()
```

```python
# Draw a vertical (|) line for the mean & try to interpret the results
mean_val = np.mean(ds)
plt.hist(ds,edgecolor='black')
plt.axvline(
    mean_val,
    color='red',
    linestyle='--'
)
plt.title("Histogram of Generated Normal Distributoin")
plt.xlabel("Score")
plt.ylabel("Frequency")
```

## Q2. For the Seaborn 'penguins' dataset, do the following: 


* Plot flipper_length_mm vs body_mass_g
* Color by species

```python
penguins = sns.load_dataset('penguins')

sns.histplot(
    data = penguins,
    x ='flipper_length_mm',
    y ='body_mass_g',
    hue = 'species'
)
plt.title("Flipper Length VS Body Mass (mm/g)")
plt.xlabel("Flipper Length")
plt.ylabel("Body Mass")
```

## Q3. For the Seaborn 'tips' dataset, do the following: 

* Create a box plot of 'total_bill' for each day
* Add hue ="sex"
* Identify which day has the highest median bill
* Comment on outliers

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(
    data = tips,
    x ='day',
    y ='total_bill',
    hue ="sex"
)
plt.title("Distribution of Total Bill by Day and Sex")
plt.xlabel("Day")
plt.ylabel("Total_Bill")
print("Saturday Has Higest Median Bill")
print("Comment on Outliners:\nObservations in")
```

## Q4. For the Seaborn 'tips' dataset, do the following:

* Create a scatter plot of total_bill vs tip.
* Color points by sex
* Use different markers for smoker
* Add a title and axis labels 
* Analyze: Do smokers tend to tip differently?

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',
    style='smoker'
)

plt.title("Distribution of smokers bills vs tips")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
```
