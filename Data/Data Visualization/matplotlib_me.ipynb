```python
# Real data

# oscar winner movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

# plot
plt.plot(years, oscar_revenue)
plt.title("Oscar Movies Revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $ Mn)")
plt.show()
```

```python
# Non Oscar Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M


plt.plot(years, oscar_revenue, color ="#FF69B4",marker='o', linestyle='-.',linewidth=4, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, '<-y', label="Non-Oscar Movies")
plt.legend(loc="upper right")
plt.title("Oscar Movies VS Non-Oscar Movies Revenue each")
plt.xlabel("Years")
plt.ylabel("Revenue (in $ Mn)")
plt.show()
```

```python
plt.plot(years, oscar_revenue, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")
plt.title("Oscar Movies VS Non-Oscar Movies Revenue each")
plt.xlabel("Years")
plt.ylabel("Revenue (in $ Mn)")
plt.legend(loc="upper right")

plt.style.available
plt.style.use("dark_background")

plt.show()
```

```python
plt.plot(years, oscar_revenue, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")
plt.title("Oscar Movies VS Non-Oscar Movies Revenue each")
plt.xlabel("Years")
plt.ylabel("Revenue (in $ Mn)")
plt.legend(loc="upper right")
plt.style.available
plt.style.use("default")

plt.show()
```

```python
plt.plot(years, oscar_revenue, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")
plt.title("Oscar Movies VS Non-Oscar Movies Revenue each")
plt.xlabel("Years")
plt.ylabel("Revenue (in $ Mn)")
plt.legend(loc="upper right")

plt.grid()
plt.show()
```

```python
with plt.xkcd():

    plt.plot(years, oscar_revenue, label="Oscar Movies")
    plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")
    plt.title("Oscar Movies VS Non-Oscar Movies Revenue each")
    plt.xlabel("Years")
    plt.ylabel("Revenue (in $ Mn)")
    plt.legend(loc="upper right")
    
    plt.tight_layout()
    plt.show()
```

```python
# movie vs revenue
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232] # in $M

plt.plot(movies, revenue)
plt.title("Movies Revenue")
plt.xlabel("Movie")
plt.ylabel("Revenue ($ M)")
plt.tight_layout()
```

```python
plt.plot(movies, revenue)
plt.title("Movies Revenue")
plt.xlabel("Movie")
plt.ylabel("Revenue ($ M)")
plt.tight_layout()
plt.savefig("final_plot.png")
```
