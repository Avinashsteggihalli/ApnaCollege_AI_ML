```python
# distribution plots (histogram)

penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.histplot(
    data=penguins,
    x='body_mass_g',
    bins=20
)
```

```python
# matrix plot
# heatmap

flight_pivot = flights.pivot(index="month",columns="year",values="passengers")

sns.heatmap(
    flight_pivot,
    cmap="coolwarm",
    annot=True,
    fmt='d'
)

plt.title("Passengers Heatmap")
plt.tight_layout()
```

```python
# best practice
fig, ax = plt.subplots() # Matplotlib
tips = sns.load_dataset("tips")
sns.lineplot( # seaborn
    data=tips,
    x='day',
    y='total_bill',
    hue='sex', # generate legend( categarise)
    marker="o", # to mak "o" syybmol marker in the line og plot
    ax=ax, # to chose the axis of the subplot
    errorbar=None # remove the shady area(remov the CI[confidence interal])
)

ax.set_title("Avg bill value for each day") # Matplotlib
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill")
fig.tight_layout()
    
```
