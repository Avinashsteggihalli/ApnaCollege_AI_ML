* from "Quotes" website , extract all quotes & authors from all pages with the life tag.

```python
import pandas as pd
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
# fetch data and store it
page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("No more valid page anymore!!..Stopping")
        break

    with open(f"scraped_data/quote_page{page_count}.html", "w",encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded from page no.-{page_count}")

    page_count = page_count + 1
```

```python
# Extract usefull info

# if line tag in each the quote then exract quote and autor of that 
# perticuler of all pages(10)
life_quotes = []
page_count = 1
total_pages = 10
while True: # open all pages and read
    if page_count == total_pages+1:
        print("All pages Completed")
        break
    with open(f"scraped_data/quote_page{page_count}.html", "r",encoding="utf-8") as f:
        html_content = f.read()
    soup = BeautifulSoup(html_content, 'lxml')
        
    all_quotes = soup.select("div.quote")
        
    for q in all_quotes:
        all_tags = []
        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())
    
        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes.append([text, author])
    print(f"Completed Page{page_count}")
    page_count = page_count + 1
print(life_quotes)
```

```python
df = pd.DataFrame(life_quotes, columns=['Quote','Author'])
df
# the below is all quotes & authors from all pages of "Quotes" webpage with the life tag.
```
