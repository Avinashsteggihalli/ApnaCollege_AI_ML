## 1. Numpy array are faster than list(optimized in C)

```python
# execution time numpy vs python list
size = 1_000_0000

try:
    #python list
    if size <= 1_000_0000:
        py_list = list(range(size))  
    else:
        print("Too much size!! to handle")
    start_time = time.time()
    sq_list = [x**2 for x in py_list]
    end_time = time.time()
    #np array
    if size <= 1_000_0000:
        np_array = np.array(range(size))  
    else:
        print("Too much size!! to handle")
    start_time2 = time.time()
    # vectorisation
    sq_array = np_array ** 2
    end_time2 = time.time()
    np_time = round(end_time2 - start_time2, 4)
    list_time = round(end_time - start_time , 4)
    print(f"Numpy Array time = {np_time} seconds")
    
    print(f"Python List time = {list_time} seconds")
    speed = "Numpy Array is speed" if np_time < list_time else "Python List is speed"
    print(speed)
except TypeError:
    print("Give size range less than 1_00_0000")
```
