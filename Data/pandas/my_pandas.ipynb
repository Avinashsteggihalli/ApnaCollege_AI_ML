```python
# Properties
# 1. Homogenous
s = pd.Series([1,2,3,4,"abc"], index=["Avinash","Charlie","Sheldon","Lenord",123])
print(s)
```

```python
# 2. Vectorized Operations
s1 = pd.Series([1,2,3,4,5,6,7])
s2 = pd.Series([5,6,7,8,9,10,11])
print(s1 + s2)
```

```python
# Handle Mising Value with NaN
s1 = pd.Series([1,2,3,4])
s2 = pd.Series([5,6,7,8,9,10,11])
print(s1 + s2)
```

```python
# Mutable values, immutable size
s1 = pd.Series([1,2,3,4])
print(s1) # origial series
s1[0] = 100
changed_s = s1.drop(0)
print(s1) # after 
print(changed_s) # changed series droping 1 index
```

```python
# filtering data

df[df["aqi"] > 100] # single condition
df[(df["aqi"] > 100) & (df["temperature"] > 30)]  
# for above multiple condition { & = ampercent = and operator; | = pipe operator = or; ~ = tilt operator = not}
df[(df["aqi"] > 100) & (df["temperature"] < 30)] 

#chaining
aqi_data = df[(df["aqi"] > 100) & (df["temperature"] > 30)] [["city","aqi"]]

aqi_data.iloc[0] # 1st value in df
#aqi_data.loc[0] - this will get error
aqi_data.loc[6] # this is correct way
```

## Query Method (Important)

```python
# Query Method
aqi_val = 100
df.query("aqi > @aqi_val & temperature > 30")[["city","aqi"]]
```
