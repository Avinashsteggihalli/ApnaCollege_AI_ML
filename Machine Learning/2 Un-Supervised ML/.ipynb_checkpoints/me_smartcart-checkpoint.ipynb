# Import Dependencies

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.decomposition import PCA
from kneed import KneeLocator
from sklearn.metrics import silhouette_score
```

# Load Dataset

```python
df = pd.read_csv("smartcart_customers.csv")
```

# View for Missing values / Errors / NUll values etc.

```python
df.head()
```

```python
df.info()
```

```python
df.info()
```

# Handle Missing Values

```python
df["Income"] = df["Income"].fillna(df["Income"].mean())
```

```python
# Feature Engineering
```

```python
df["Education"].value_counts()
```

```python
df["Education"] = df["Education"].replace({
    "Graduation" : "Graduate" , "PhD" : "Postgraduate",
    "Master" : "Postgraduate" , "2n Cycle" : "Undergraduate",
    "Basic" : "Undergraduate"})
```

```python
df["Education"].value_counts()
```

```python
df["Age"] = 2026 - (df["Year_Birth"])
```

```python
df["Age"].head()
```

```python
df["Marital_Status"].value_counts()
```

```python
df["Living_with"] = df["Marital_Status"].replace({
    "Married" : "Partner" , "Together" : "Partner",
    "Single" : "Alone" , "Divorced" : "Alone" ,
    "Widow" : "Alone" , "Absurd" : "Alone", "YOLO" : "Alone"})
```

```python
df["Living_with"].value_counts()
```

```python
df["Total_Spending"] = df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] +  df["MntSweetProducts"] + df["MntGoldProds"] + df["MntFishProducts"]
```

```python
df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"] , format="%d-%m-%Y")

baseline_year = df["Dt_Customer"].max()

df["Customer_Tenure_Days"] = (baseline_year - df["Dt_Customer"]).dt.days
```

```python
df.info()
```

```python
# Drop columns
```

```python
cols = ["ID" ,"Year_Birth", "Marital_Status", "Kidhome", "Teenhome", "Dt_Customer"]
spending_cols = ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"]

total_cols = cols + spending_cols
df_cleaned = df.drop(columns=total_cols)
# print(total_cols)
```

```python
df_cleaned.head()
```

# Handle Outliers

```python
cols1 = ["Income", "Age", "Recency", "Response", "Total_Spending", "Total_Children"]

sns.pairplot(df_cleaned[cols1])
```

```python
# len_df = len(df_cleaned)
df_before = len(df_cleaned)
df_cleaned = df_cleaned[(df_cleaned["Age"] < 90)]
df_cleaned = df_cleaned[(df_cleaned["Income"] < 600_000)]
df_after = len(df_cleaned)

print("Before Removeing Outlier:", df_before)
print("After Removeing Outlier:", df_after)
```

```python
# HeatMap
```

```python
corr_matrix = df_cleaned.corr(numeric_only=True)

plt.figure(figsize=(8,6))
sns.heatmap(
    corr_matrix,
    annot=True,
    annot_kws={"size":6},
    cmap="coolwarm"
)
```

```python
# Encode
```

```python
ohe = OneHotEncoder()
cat_cols = ["Education", "Living_with"]
encod_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
encod_df = pd.DataFrame(encod_cols.toarray(), columns=ohe.get_feature_names_out(cat_cols) , index=df_cleaned.index)
```

```python
encod_df.head()
```

```python
df_encoded = pd.concat([df_cleaned.drop(columns=cat_cols), encod_df], axis=1)
```

```python
df_encoded.head()
```

```python
# Scaling
```

```python
scaler = StandardScaler()
X = df_encoded
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

# Dimension Reduction

```python
#2D, want to see in 3d, so n_components=3
pca = PCA(n_components=3)

X_pca = pca.fit_transform(X_scaled)
```

```python
fig = plt.figure(figsize=(8,6))

axe = fig.add_subplot(111, projection="3d")
axe.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2])

axe.set_xlabel("PC1")
axe.set_ylabel("PC2")
axe.set_zlabel("PC3")

axe.set_title("X_pca 3D PLOT")
```

# Optimal K- value

# Elbow

```python
wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(1,11), wcss, marker="o")
```

```python
knee = KneeLocator(range(1,11), wcss, curve="convex", direction="decreasing")

print("Optimal K:", knee.knee)
```

# Silhouette Score

```python
ss = []
for k in range(2,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca, labels)

    ss.append(score)
```

```python
plt.plot(range(2,11), ss, color="red", marker="x")
```

```python
# both graph in one
```

```python
k_range = range(2,11)

fig, ax1 = plt.subplots(figsize=(8,6))

ax1.plot(k_range, wcss[:len(k_range)], marker="o", color="blue")
ax1.set_xlabel("K values")
ax1.set_ylabel("WCSS")

ax2 = ax1.twinx()
ax2.plot(k_range, ss[:len(k_range)], marker="x", color="red")
ax2.set_ylabel("SS")
```

# Kmeans

```python
kmeans = KMeans(n_clusters=4, random_state=42)
labels_kmeans = kmeans.fit_predict(X_pca)
```

# 3d Visualization

```python
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2], c=labels_kmeans)
```

# Agglomerative Clustering (Hierarchical Clustering)

```python
agg_clu = AgglomerativeClustering(n_clusters=4, linkage="ward")

labels_agg = agg_clu.fit_predict(X_pca)
```

# 3d Visualization

```python
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2], c=labels_agg)
```

# DBSCAN

```python
dbscan = DBSCAN()

labels_dbscan = dbscan.fit_predict(X_pca)
```

```python
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2], c=labels_dbscan)
```
