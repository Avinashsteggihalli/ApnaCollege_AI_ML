```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("thyroid_dataset.csv")
```

```python
df.head()
```

```python
X = df.drop("Outlier_label", axis=1)
```

```python
#scale X
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
isf = IsolationForest(n_estimators=200, contamination="auto", random_state=42)
```

```python
labels = isf.fit_predict(X_scaled)
```

```python
pca = PCA(n_components=2, random_state=42)

X_pca = pca.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(x=X_pca[:,0], y=X_pca[:,1], c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
```

```python
num_outliners = np.sum(labels == -1)
num_normal = np.sum(labels == 1)

print("Outliners:", num_outliners)
print("Normal:", num_normal)
```

# LOF

```python
lof = LocalOutlierFactor(contamination=0.036)

labels = lof.fit_predict(X_scaled)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(x=X_pca[:,0], y=X_pca[:,1], c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
```

```python
num_outliners = np.sum(labels == -1)
num_normal = np.sum(labels == 1)

print("Outliners:", num_outliners)
print("Normal:", num_normal)
```
