### 2. Take two numbers as input from the user and print their sum, difference, product, and quotient.

(note: 10 / 5 = 2)

e.g:-

10 = dividend

5  = divisor

2  = *quotient* 

```python
num1 = int(input("Enter a number(num1): "))
num2 = int(input("Enter a number(num2): "))
print("Num1 is: ",num1,"\nNum2 is: ",num2)
sum = num1 + num2
print("\nSum of two numbers is: ",sum)
diff = num1 - num2
print("\nDifference of two numbers is: ",diff)
prdt = num1 * num2
print("\nProduct of two numbers is: ",prdt)
qut = num1 / num2
print("\nQuotient of two numbers is: ",qut)
```

### 3. Ask the user to enter two integers and one float. Convert them all to floats and print their average.

```python
num1 = int(input("Enter a integer number: "))
num2 = int(input("Enter a integer number: "))
num3 = float(input("Enter a Float number: "))

print(type(num1), type(num2), type(num3))

num1_f = float(num1)
num2_f = float(num2)
num3_f = float(num3)

print(type(num1_f), type(num2_f), type(num3_f))

avg = (num1_f + num2_f + num3_f)/3

print("Average of the numbers is : ", avg)
```

### 4. The user enters a string containig a number (e.g., "45" ). Convert it to :
* an integer
* a float
* a string again

Print all three values with thier types.


```python
str_num = input("Enter a number(which will be a string): ")

print("The entered string(number):", str_num, type(str_num))
# to integer
a = int(str_num)
print("String to integer: ", a, type(a))

# to float
b = float(a)
print("Integer to Float: ", b, type(b))

# to string
c = str(b)
print("Again to String: ", c, type(c))
```

### 6. Write a program to swap values of two numbers entered by the user.

```python
a = input("Enter value for 'a' : ")
b = input("Enter a value for 'b' : ")
print("Before Swap-")
print("Value of 'a' is: ",a)
print("Value of 'b' is: ",b)
a, b = b, a
print("\nAfter Swap-")
print("Value of 'a' is: ",a)
print("Value of 'b' is: ",b)
```

### 7. Ask the user for the Temperature in Celcius (string input). Convert it to float , then calculate and print them to Fahrenheit.

 Conversion Formula : FahrenheitTemp =  (CelciusTemp * (9/5)) + 32

```python
#temp input in string
c_temp = input("Enter the Temperature in Celcius: ")
# c_temp value
print("Entered Celcius value: ",c_temp)
# convert to float
c_temp_f = float(c_temp)
# celcius to fehrenheit
feh_temp = (c_temp_f * (9/5)) + 32
print("\nThe Fahrenheit Value is: ", feh_temp)
```

### 8. Take the radius (r) as user input and print the area.

```python
radius = float(input("Enter the radius value: "))
print("Given Radius: ",radius)
PI = 3.142
area = PI * (radius ** 2)
print("\nArea of circle is : ", area)
```

### 9. Ask the user for: Principal (P), Rate (R), Time (T). Convert all to float and compute simple interest:

SI = (P * R * T)/ 100

```python
P = float(input("Enter the Principal value: "))
R = float(input("Enter the Rate value: "))
T = float(input("Enter the Time value: "))
print("Principal : ", P,"\nRate : ", R, "\nTime : ",T)
SI = (P * R * T )/100
print("Simple Interest is: ",SI)
```

### 10. Take a decimal number as input (like 45.78) and output its:-

* integer part - 45
* fraction part - .78

```python
deci_num = float(input("Enter a float(decimal) value: "))
print("Entered value: ", deci_num)
int_part = int(deci_num)
fract_part = deci_num - int(deci_num)
print("\nInteger part- ",int_part)
print("Fraction part- ",fract_part)
```
