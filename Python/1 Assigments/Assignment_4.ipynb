### 1. Create A BankAccount class with attribute account_number, owner_name, and balance.

Add methods to deposit. withdraw and check balance. 

```python
class BankAccount:
    def __init__(self, account_number, owner_name, balance):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return f"Deposited Amount: {amount} .Successfully Money Deposited!!" , self.check_bal()
    def withdraw(self, amount):
        self.balance -= amount
        return f"Withdrawn Amount: {amount} .Successfully Money Withdrawn!!" , self.check_bal()
    def check_bal(self):
        return f"Remaining Amount: {self.balance}"
bank_acc = BankAccount(1234, "Avinash", 21000)
print("Account no.:", bank_acc.account_number, "; Owner Name:", bank_acc.owner_name,"; My account Balance:", bank_acc.balance)
print("\n")
print(bank_acc.deposit(1000))
print(bank_acc.withdraw(5000))
print(bank_acc.check_bal())
```

### | Concept: Classes & Object
### 2. Create a class "Book" with the following attributes:

* title
* author
* list of review 

Add add methods to:

* add a new review
* count review
* display all reviews

```python
class Book:
    def __init__(self, title, author, list_review):
        self.title = title
        self.author = author
        self.list_review = list_review

    def add_new_review(self, new_reiew):
        self.list_review.append(new_reiew)
        return "New review added: ", new_reiew
    def count_reiew(self):
        return "Number of Reviews: ",len(self.list_review)
    def display_all_review(self):
        return "All review: ",self.list_review

book = Book("Everything_IS_Everything", "Avinash S Teggihalli", ["So awesome!!","Wonderfull book...", \
                                                                 "Must read book", "Little bit costly"])
print("Title of Book :",book.title,"; Author name: ",book.author,"; \nList of reviews: ",book.list_review)
print("\n")
print(book.add_new_review("!!Most valuable book!!"))
print(book.count_reiew())
print(book.display_all_review())
```
