### 1. Create A BankAccount class with attribute account_number, owner_name, and balance.

Add methods to deposit. withdraw and check balance. 

```python
class BankAccount:
    def __init__(self, account_number, owner_name, balance):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return f"Deposited Amount: {amount} .Successfully Money Deposited!!" , self.check_bal()
    def withdraw(self, amount):
        self.balance -= amount
        return f"Withdrawn Amount: {amount} .Successfully Money Withdrawn!!" , self.check_bal()
    def check_bal(self):
        return f"Remaining Amount: {self.balance}"
bank_acc = BankAccount(1234, "Avinash", 21000)
print("Account no.:", bank_acc.account_number, "; Owner Name:", bank_acc.owner_name,"; My account Balance:", bank_acc.balance)
print("\n")
print(bank_acc.deposit(1000))
print(bank_acc.withdraw(5000))
print(bank_acc.check_bal())
```

### | Concept: Function Overriding
### 4. Create a class 'Shape' with a method area(). Create subclass Circle, Rectangle, and Triangle that **Override** the area() method.

```python
class Shape:
    PI = 3.142
    def area(self):
        print("Parent class: Shape")
        print("pi value:", self.PI)
class Circle(Shape):
    def area(self, radius):
        print("Area of Circle: PI*r**2: ", self.PI*(radius**2))
        

class Rectangle(Shape):
    def area(self,length, breadth):
        print("Area of Rectangle: l * b: ", length*breadth)

class Triangle(Shape):
    def area(self, breadth, height):
        print("Area of Triangle: 1/2(b*h): ", 1/2*(breadth*height))


s = Shape()
circle = Circle()
rectangle = Rectangle()
triangle = Triangle()

s.area()
circle.area(4)
rectangle.area(5,6)
triangle.area(7,8)
```

### | Concept: Inheritance
### 5. Create a **base** class 'Vehicle' with attributes like brand and model. Create two **subclass** 'Car'  and 'Bike' that add extra attributes - seat (in Car) & engine_cc (in Bike). 

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def get_info(self):
        print("Vehicle: ", self.brand, self.model)

class Car(Vehicle):
    def __init__(self, brand, model, seat):
        self.seat = seat
        super().__init__(brand, model)

    def get_info(self):
        super().get_info()
        print("Type Car: seat-", self.seat)

class Bike(Vehicle):
    def __init__(self, brand, model, engine_cc):
        self.engine_cc = engine_cc
        super().__init__(brand, model)

    def get_info(self):
        super().get_info()
        print("Type Bike: Engine-", self.engine_cc)

car = Car("TATA", "Punch", seat="Leather")
bike = Bike("Hero", "Hf-Delux", engine_cc="100cc")

print("-------Car--------")
car.get_info()
print("\n")
print("-------Bike--------")
bike.get_info()


```

### | Concept: Abstraction
### 6. Create a abstract class 'Employee' with an **abstarct method** calculate_salary(). Create subclasses Intern, FullTimeEmployee, and ContractEmployee that implement the method differently.

```python
from abc import ABC, abstractmethod

@abstractmethod
class Employee(ABC):
    def calculate_salary():
        pass

class Intern(Employee):
    def calculate_salary(self):
        print("Inern's Salary Calculated")

class FullTimeEmployee(Employee):
    def calculate_salary(self):
        print("Full Time Employee's Salary Calculated")

class ContractEmployee(Employee):
    def calculate_salary(self):
        print("Contract Employee's Salary Calculated")

intern = Intern()
full_time = FullTimeEmployee()
contract = ContractEmployee()

intern.calculate_salary()
full_time.calculate_salary()
contract.calculate_salary()
```

### | Concept: Multiple Inheritance 
### 9. Create the following classes: Herbivore, Carnivore, Omnivore with some attributes & method. Then create a class 'Bear' that inherits from all the above classes to showcase how the multiple inheritance works.

```python

class Herbivore:#  only plants
    def __init__(self, name):
        self.name = name

    def eat_plant(self):
        print(f"{self.name} the Bear is eating plant!!")
class Carnivore:# only animals
    def __init__(self, name):
        self.name = name

    def eat_animal(self):
        print(f"{self.name} the Bear is eating animals!!")
class Omnivore: # both animal and plants
    def __init__(self, name):
        self.name = name

    def eat_aniplant(self):
        print(f"{self.name} the Bear is eating both plants and animal!!")

class Bear(Herbivore, Carnivore, Omnivore):
    def __init__(self, name):
        self.name = name
        Herbivore.__init__(self, name)
        Carnivore.__init__(self, name)
        Omnivore.__init__(self, name)
        

my_bear = Bear("Grizzy")

my_bear.eat_plant() , my_bear.eat_animal(), my_bear.eat_aniplant()
```

### 10. Mini Project - OOP Chat System
Let's create a Chat System using OOPs concepts. We have to create classes:
* user
* Messege
* ChatRoom

And we have to implement functions:

* sending messeges
* viewing chat history
* user joining and leave the chatroom

```python
# -------------------------------
# Message Class
# -------------------------------
class Message:
    message_counter = 1   # simple counter
    def __init__(self, sender, content):
        self.sender = sender
        self.content = content
        self.id = Message.message_counter
        Message.message_counter == 1
    def __str__(self):
        return f"({self.id}) {self.sender.username}: {self.content}"
# -------------------------------
# User Class
# -------------------------------
class User:
    def __init__(self, username):
        self.username = username
        self.chatroom = None
    def join_chatroom(self, chatroom):
        if self.chatroom:
            print(f"{self.username} is already in a chatroom.")
        else:
            chatroom.add_user(self)
            self.chatroom = chatroom
            print(f"{self.username} joined {chatroom.name}")
    def leave_chatroom(self):
        if not self.chatroom:
            print(f"{self.username} is not in any chatroom.")
        else:
            self.chatroom.remove_user(self)
            print(f"{self.username} left {self.chatroom.name}")
            self.chatroom = None
    def send_message(self, content):
        if not self.chatroom:
            print(f"{self.username} cannot send a message (not in a chatroom).")
        else:
            self.chatroom.broadcast(self, content)
# -------------------------------
# ChatRoom Class
# -------------------------------
class ChatRoom:
    def __init__(self, name):
        self.name = name
        self.users = []
        self.messages = []
    def add_user(self, user):
        self.users.append(user)
    def remove_user(self, user):
        self.users.remove(user)
    def broadcast(self, sender, content):
        message = Message(sender, content)
        self.messages.append(message)
        print(message)   # Show message to all users
    def show_chat_history(self):
        print(f"\nChat History of {self.name}:")
        for msg in self.messages:
            print(msg)
        print()
# ---------------------------------------
# Example Usage
# ---------------------------------------
if __name__ == "__main__":
    room = ChatRoom("Python Lounge")
    u1 = User("Alice")
    u2 = User("Bob")
    u3 = User("Charlie")
    u1.join_chatroom(room)
    u2.join_chatroom(room)
    u1.send_message("Hello everyone!")
    u2.send_message("Hi Alice!")
    u3.join_chatroom(room)
    u3.send_message("Hey guys, what's up?")
    room.show_chat_history()
    u1.leave_chatroom()
    u2.leave_chatroom()
    u3.leave_chatroom()
```
