### Q1. Create a program that:

1. Opens a file "names.txt" in write mode.
2. Writes 5 names (one per line) entered by the user 
3. Then opens  the same file read mode and prints all names

```python
open("names.txt",'x') # created new file named "name.txt"
```

```python
with open("names.txt",'w') as f:
    for i in range(5):
        user = input("Enter 5 names(one per line): ")
        f.writelines(user+"\n")
```

```python
with open("names.txt",'r') as f:
    data = f.read()
    print(data)
```

### Q3. Create a program that:

1. list of numbers: [5, 10, 15, 20, 25]
2. uses a list comprehension to create a new list with only numbers greater than 15.
3. prints the new list.

```python
list_num = [5, 10, 15, 20, 25]
new_lst = [x for x in list_num if x > 15]
print(new_lst)
```

### Q4. Create a python dictionary of cities and their populations. save it to 'cities.json'.:

1. then load the json and print each city and its population.
2. ask the user for a new city & its populaion- update this info in the json file.

```python
import json #importing py json module
```

```python
# dictionary creation
py_dict = {"Haveri":25_000,"Sirsi":17_000,"Bagalkot":40_000,"Dhrawad":60_000,"Bangaluru":1_00_000}
```

```python
with open("cities.json",'x') as f: # to create new file
    json.dump(py_dict, f, indent=4)
```

```python
user_city = input("Enter city name: ") # city input
user_pop = int(input("Enter its population: ")) # population input
py_dict[user_city] = user_pop # append new key-value pair in existing dict
with open("cities.json",'w') as f: # open the file as f with 'w' mode with auto close keyword
    json.dump(py_dict, f, indent=4) #overwrite te file with updated data
```

### Q5. Write a program that tries to open "data.txt" in read mode. if the file does not exits, catch the exception and print "File not found!".



```python
try:
    with open("data.txt",'r') as f:
        f.read()
        print("File opened!!")

except FileNotFoundError:
    print("File Not Found!!")
```
