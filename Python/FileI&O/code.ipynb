practice problem

find a word "python" from a "test.txt" file if it exists then print(the word, line number)

```python
with open("test.txt", "x") as f: # to create a test.txt file
    f.write("We are using a language\nfor coding\ni.e: python")
```

```python
f.close() # closed it
```

```python
data = True # to continue to while loop
line = 0 # to count the nth line word
word = "coding" # insert word here

with open("test.txt",'r') as f: # to close the file automattically
    while data:
        line+=1
        data = f.readline() # read line by line
        if word in data:
            print(f"{word} is in line {line}.")

```
