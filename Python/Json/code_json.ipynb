```python
import json
```

```python
js_str = '{"name": "Avinash","isStudent": true,"adderess":{"city":"Haveri","country":"India"},"subjects": ["Python","AI/ML"]}'
```

```python
py_obj = {
    "name":"Avinash S Teggihalli",
    "Compay":"Wipro",
    "Join Date": "27, May 2025",
    "isStudent" : True
}
```

## loads()

json ----> python (string)

```python
# load -Javascript_str to Pyhon Object(i.e Dict)
pyt_obj = json.loads(js_str) # in .loads() - 's' is for string format, not plural
```

```python
print(pyt_obj) # now it is converted to py_object
print(type(pyt_obj))
```

## dumps()

python ---> json (string)

```python
# dump - py_obj to javas_str
javas_str = json.dumps(py_obj) # in .dumps() - 's' is for string format, not plural
print(javas_str)
print(type(javas_str))
```

## load() , without 's' , ie. while dealing with files

json --> python (in file) to read file from json format

```python
with open("data.json","r") as f: # load -Javascript_str file to Pyhon Object file(i.e Dict) - just to read
    pyt_obj = json.load(f) # loaded file has to be assigned to a variable to print
```

## dump()

python ---> json (in file) ; to write json to a file

```python
with open("data.json","w") as f:
    json.dump(py_obj, f, indent=4, sort_keys=True) # assign to argument (the file that has to dump, the dumped place) , (a ,b) - a to b
    # indent - with make it as proper indent(space , alignment)
    # sort_key - will sort the key in alphabetic order
```
