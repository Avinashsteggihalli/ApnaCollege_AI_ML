# Abstract Class Importance Demo
This notebook demonstrates the difference between **without abstract class** and **with abstract class** using a real bug scenario.

## ❌ 1. Without abstract class — silent bug
You want all vehicles to have `start_engine()`, but nothing forces developers to implement it.

```python
class Vehicle:
    pass

class Bike(Vehicle):
    def ride(self):
        print("Riding bike")

b = Bike()
print("Bike created successfully.")

# This will crash because start_engine() is missing
b.start_engine()
```

## ✅ 2. With abstract class — bug caught instantly
Now Python enforces the rule properly.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

class Bike(Vehicle):
    def ride(self):
        print("Riding bike")

# This will raise TypeError before program runs
b = Bike()
```

## Summary
* **Without abstract class** → program looks fine but crashes later.
* **With abstract class** → Python stops incomplete classes instantly.

Abstract classes act like **seatbelts for your code**, preventing future disasters.
