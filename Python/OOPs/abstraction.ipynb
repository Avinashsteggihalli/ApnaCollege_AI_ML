```python
from abc import ABC , abstractmethod # abstract base class

class Animal(ABC): # here ABC module is inherited to 'Animal' class
    @abstractmethod
    def make_sound(self):
        pass

class Lion(Animal):
    def make_sound(self):
        print("Rorr")

class Cow(Animal):
    def make_sound(self):
        print("Moo")

lion = Lion()
cow = Cow()
```

```python
lion.make_sound()
```

```python
cow.make_sound()
```
