# Data Handling


```python
# public/protected/private attribute ; getter & setter fuction
class Student:
    def __init__(self, name, section, roll):
        self.name = name # public : no underscore
        self._section = section # protected : 'single underscore' : no enforce : by convention
        self.__roll = roll #private : 'double underscore' ; data mangling ; little enforce

    def get_roll(self): # 'getter funtion' - to access private attribute 
        return self.__roll # put the same as called in constructor
    
    def set_roll(self, newRoll): 
        self.__roll = newRoll
        return self.__roll
    
std = Student("Avinash", "B", "2024wd15148")
print(std.name) #public
```

```python
print(std._section) # protected
```

```python
print(std.__roll) # private
```

```python
print(std.get_roll()) #calling 'getter' function to access private attribute
```

```python
print(std.set_roll("120pme019")) #setting newValue to the private attribute
```

```python
print(std.get_roll()) # newRoll is been set from 'setter' function and called from 'getter' function
```
