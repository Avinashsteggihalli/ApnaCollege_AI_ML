# Inheritance

* single level Inherit
* multi level Inherit

```python
class Employee: # parent class (base)
    start_time = "10am" # both start and end time are common for all staff
    end_time = "6pm"

    def change_end_time(self, new_end_time):
        self.end_time = new_end_time

class Teacher(Employee): # inherited from the parent class attr (inheritance)
    def __init__(self, subject): # Child class
        self.subject = subject

class AdminStaff(Employee): # inherited from the parent class attr (inheritance)
    def __init__(self, role): # child class
        self.role = role

t1 = Teacher("AI") # assigned str to subject instance attr
```

```python
t1.subject # calling object 'subject' of teacher 't1'
```

```python
t1.start_time, t1.end_time # initial start and end time ( 10pm to 6pm)
```

```python
t1.change_end_time("5pm") # changed new end time 
```

```python
t1.start_time, t1.end_time # changed values i.e 5pm
```

```python
staff1 = AdminStaff("Manager") # assigned to str to 'Adminstaff' child class's 'role' in 'staff1' object
```

```python
staff1.role , staff1.start_time, staff1.end_time
```

## till now Single level inheritance

now below are **Multilevel**

grand parent >> parent >> child

```python
class Employee: # Grand parent class (base)
    start_time = "10am" # both start and end time are common for all staff
    end_time = "6pm"

class AdminStaff(Employee):  # parent class
    def __init__(self, role): 
        self.role = role

class Accountant(AdminStaff): # child class ; take the parent+grand parent class properties
    def __init__(self,salary, role):
        super().__init__(role) # to invoke the constructer of parent class/grand parent {but in this parent class} 
        self.salary = salary


acc1 = Accountant(30_000, "Account Manager")
```

```python
acc1.role, acc1.salary, acc1.start_time, acc1.end_time
```

## Now Multiple level Inheritance

```python
class Teacher: # parent 1
    def __init__(self, salary):
        self.salary = salary

class Student: # parent 2
    def __init__(self, gpa):
        self.gpa = gpa

class TA(Teacher, Student):
    def __init__(self, name, salary, gpa):
        self.name = name
        super().__init__(salary) # call parent 1 constructor (Teacher)
        Student.__init__(self, gpa) # call parent 2 constructor (Student)

ta1 = TA("Ainash", 40_000, 9.4)
```

```python
ta1.name, ta1.salary, ta1.gpa
```
