# Inheritance

```python
class Employee: # parent class (base)
    start_time = "10am"
    end_time = "6pm"

    def change_end_time(self, new_end_time):
        self.end_time = new_end_time

class Teacher(Employee): # inherited the parent class attr (inheritance)
    def __init__(self, subject): # Child class
        self.subject = subject

t1 = Teacher("AI") # assigned str to subject instance attr
```

```python
t1.subject # calling object 'subject' of teacher 't1'
```

```python
t1.start_time, t1.end_time # initial start and end time ( 10pm to 6pm)
```

```python
t1.change_end_time("5pm") # changed new end time 
```

```python
t1.start_time, t1.end_time # changed values i.e 5pm
```
