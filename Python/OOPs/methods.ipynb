```python
class Laptop:
    storage_type = "ssd"

    def __init__(self , RAM, storage):
        self.RAM = RAM
        self.storage = storage
    
    @classmethod # decorator 
    def get_storage_type(cls): #class method
        print(cls.storage_type)
        #print(cls.storage_type ,cls.RAM) # it is trying to find 'RAM' in class function as we have put @classmethod decorator

    def get_info(self): #instance method
        print(f"Laptop has {self.RAM} RAM & {self.storage} Storage")

    @staticmethod
    def calc_discount(price, discount):
        final_price = price - (discount * price / 100)
        print(f"Final price = {final_price}")

lap1 = Laptop("16gb", "512gb")
lap2 = Laptop("8gb", "256gb")


# print((lap1.storage_type)+" type") # directly calling outside
```

```python
# instance method
lap1.get_info() 
lap2.get_info() 
```

```python
lap1.get_storage_type() # calling from class method
```

```python
# static method
Laptop.calc_discount(100, 10)
```

# product store

design and create a online store for Product(name, price)

Track total products being created

Create a static method to calculate discount on each product based on a % parameter.


```python
class Product:
    count = 0
    
    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.count += 1

    def get_info(self):
        print(f"Price of {self.name} is Rs.{self.price}")

    @classmethod
    def get_count(cls):
        print(f"Total number of Products: {cls.count}")

    @staticmethod
    def clac_discount(price, discount_price):
        final_price = price - (discount_price * price / 100)
        print(f"Discount percentage is {discount_price}%")
        print(f"Final price is {final_price}")

p1 = Product("Phone", 10_000)
p2 = Product("Laptop", 30_000)
p3 = Product("TV", 40_000)

```

```python
Product.clac_discount(p1.price, 5) # discount for product1
```
