## Function Overriding

```python
class Employee:
    def get_deignation(self):
        print("Designation = Employee")

class Teacher(Employee):
    def get_deignation(self):
        print("Designation = Teacher")

d1 = Teacher()
```

```python
d1.get_deignation() # perticularly we called teacher class
```

```python
d2 = Employee()
d2.get_deignation() # we called emploee now
```

## Duck Typing

which walk like duck & which quack like a duck

```python
class Employee():
    def get_designation(self):
        print("Designation = Employee")

class Teacher():
    def get_designation(self):
        print("Designation = Teacher")
```

```python
emp = Employee()
emp.get_designation()
```

```python
t = Teacher()
t.get_designation()
```

# this is best example of poly-morphism

```python
class Duck: #animal 1
    def quack(self):
        print("Quack!")

class Person:
    def quack(self): # animal 2
        print("I'm mimicking a duck!")

def make_it_quack(animal): # created method to call perticuler class (animal1 or animal 2) ,
    animal.quack()         #here Animal is comman/single and duck ,person are different forms
    

d = Duck()
p = Person()

make_it_quack(d)  # Output: Quack!
make_it_quack(p)  # Output: I'm mimicking a duck!
```
