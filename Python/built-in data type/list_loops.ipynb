## Linear search

```python
import time
start_time = time.time()
l = [1, 3, 6, 10, 4, 100]

val = 10
indx = 0
for i in l:
    if (i == val):
        print(f"Found the number {val} in the index {indx}")
        break
    indx+=1
end_time = time.time()
elapsed_time = end_time - start_time
print(f"Execution time: {elapsed_time:.4f} seconds")
```

```python
print(l.index(100)) # to find the position of the perticuler element
```
