```python
info1 = [
    ("Alice","Math"),
    ("Bob", "Science"),
    ("Alice" , "Science"),
    ("Charlie", "Math"),
    ("Bob", "Math"),
    ("Alice", "English"),
    ("Charlie", "English")
]
info1
```

```python
dict = {} # dictionary , variable , equal , curly braces

for name, course in info1: # for loop, membership operator, list(paired(tuples))
    if dict.get(name) == None: # conditonal statement, .get() method , parenthesis , intentation, equal to, None, colon
        dict.update({name : set()}) # .update() method , keys : value pair
        dict[name].add(course)
    else:
        dict[name].add(course)
dict # calling the dictionary
```
