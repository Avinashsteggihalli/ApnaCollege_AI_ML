```python
try:
    n = int(input("Enter a number: "))
    a = 10/n

except ZeroDivisionError:
    print(f"ZeroDivisionError: Divide by zero 0 is not allowed")

except ValueError:
    print("ValueError: Invalid input")
else:
    print("a=",a)
```
