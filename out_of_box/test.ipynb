```python
import calendar

def func(mm,dd,yyyy):
    #enter code here
    day = calendar.weekday(yyyy, mm, dd)

    match day:
        case 0:
            return "MONDAY"
        case 1:
            return "TUESDAY"
        case 2:
            return "WEDNESDAY"
        case 3:
            return "THURSDAY"
        case 4:
            return "FRIDAY"
        case 5:
            return "SATURDAY"
        case 6:
            return "SUNDAY"
        case _:
            return "Enter proper format"

    
        
if __name__ == "__main__":
    #call func here
    mm,dd,yyyy = map(int, input().strip().split())
    result = func(mm,dd,yyyy)
    print(result)
```
